```python
import numpy as np

states = [0,1,2]
actions = [0,1]
gamma = 0.9

P = {

0:{
0:[(1,1,5)],
1:[(1,2,10)]
},

1:{
0:[(1,0,-1)],
1:[(1,2,2)]
},

2:{
0:[(1,2,0)],
1:[(1,2,0)]
}

}

# VALUE ITERATION

print("VALUE ITERATION\n")

V=np.zeros(len(states))

for iteration in range(10):

    new_V=np.zeros(len(states))

    for s in states:

        action_values=[]

        for a in actions:

            value=0

            for prob,next_state,reward in P[s][a]:
                value += prob*(reward + gamma*V[next_state])

            action_values.append(value)

        new_V[s]=max(action_values)

    print("Iteration",iteration+1)

    for s in states:
        print("State",s,"=",round(new_V[s],4))

    print("----------------")

    if np.max(np.abs(new_V-V))<0.001:
        break

    V=new_V

# POLICY ITERATION

print("\nPOLICY ITERATION\n")

policy=np.zeros(len(states),dtype=int)

iteration=1

while True:

    print("Policy Iteration",iteration)

    # Policy Evaluation
    V=np.zeros(len(states))

    for i in range(20):

        for s in states:

            a=policy[s]

            prob,next_state,reward=P[s][a][0]

            V[s]=reward + gamma*V[next_state]


    print("State Values:")

    for s in states:
        print("State",s,"=",round(V[s],4))

    print("Current Policy =",policy)

    # Policy Improvement
    stable=True

    for s in states:

        old_action=policy[s]

        action_values=[]

        for a in actions:

            prob,next_state,reward=P[s][a][0]

            action_values.append(
                reward + gamma*V[next_state]
            )

        policy[s]=np.argmax(action_values)

        if old_action!=policy[s]:
            stable=False

    print("Updated Policy =",policy)

    print("-------------------")

    if stable:
        break

    iteration+=1

print("\nFinal Optimal Policy:")
print(policy)
```
