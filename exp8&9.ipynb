```python
import numpy as np

# Grid size
rows, cols = 3,3

# Parameters
gamma = 0.9
reward = -1

# Goal State
goal=(2,2)

# Actions
actions = {
'UP':(-1,0),
'DOWN':(1,0),
'LEFT':(0,-1),
'RIGHT':(0,1)
}

# Initialize values
V=np.zeros((rows,cols))

###################################
# VALUE ITERATION
###################################

for k in range(50):

    newV=np.copy(V)

    for i in range(rows):
        for j in range(cols):

            if (i,j)==goal:
                continue

            action_values=[]

            for a in actions:

                di,dj=actions[a]

                ni=i+di
                nj=j+dj

                # Boundary check
                if ni<0 or ni>=rows or nj<0 or nj>=cols:
                    ni,nj=i,j

                val = reward + gamma*V[ni][nj]

                action_values.append(val)

            newV[i][j]=max(action_values)

    if np.max(abs(newV-V))<0.001:
        break

    V=newV

print("Value Iteration Results:\n")

for i in range(rows):
    for j in range(cols):
        print("State:",(i,j),"Value:",round(V[i][j],2))


###################################
# POLICY ITERATION
###################################

# Initial random policy
policy=np.full((rows,cols),"DOWN",dtype=object)

while True:

    # Policy Evaluation
    for k in range(30):

        for i in range(rows):
            for j in range(cols):

                if (i,j)==goal:
                    continue

                act=policy[i][j]

                di,dj=actions[act]

                ni=i+di
                nj=j+dj

                if ni<0 or ni>=rows or nj<0 or nj>=cols:
                    ni,nj=i,j

                V[i][j]=reward + gamma*V[ni][nj]


    stable=True

    # Policy Improvement
    for i in range(rows):
        for j in range(cols):

            if (i,j)==goal:
                continue

            old=policy[i][j]

            best_action=None
            best_value=-9999

            for a in actions:

                di,dj=actions[a]

                ni=i+di
                nj=j+dj

                if ni<0 or ni>=rows or nj<0 or nj>=cols:
                    ni,nj=i,j

                val=reward + gamma*V[ni][nj]

                if val>best_value:
                    best_value=val
                    best_action=a

            policy[i][j]=best_action

            if old!=policy[i][j]:
                stable=False

    if stable:
        break


print("\nPolicy Iteration Results:\n")

for i in range(rows):
    for j in range(cols):

        if (i,j)==goal:
            print("State:",(i,j),"Action: GOAL, Value:",round(V[i][j],2))

        else:
            print("State:",(i,j),
                  "Action:",policy[i][j],
                  "Value:",round(V[i][j],2))
```
