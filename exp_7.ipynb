```python
import numpy as np

# Hidden states
states = ['Rainy','Sunny']

# Observation sequence
# Walk=0, Shop=1, Clean=2
obs = [0,1,2]

# Initial probabilities
pi=np.array([0.6,0.4])

# Transition probabilities
A=np.array([
[0.7,0.3],
[0.4,0.6]
])

# Emission probabilities
B=np.array([
[0.1,0.4,0.5],
[0.6,0.3,0.1]
])

T=len(obs)
N=len(states)

# -------- Forward Algorithm --------

alpha=np.zeros((T,N))

# Initialization
alpha[0]=pi*B[:,obs[0]]

# Recursion
for t in range(1,T):
    for j in range(N):
        alpha[t,j]=sum(alpha[t-1]*A[:,j])*B[j,obs[t]]

print("Forward Probabilities:")
print(alpha)

print("Observation Probability =",
sum(alpha[T-1]))


# -------- Backward Algorithm --------

beta=np.zeros((T,N))

# Initialization
beta[T-1]=1

# Recursion
for t in range(T-2,-1,-1):
    for i in range(N):
        beta[t,i]=sum(
        A[i]*B[:,obs[t+1]]*beta[t+1]
        )

print("\nBackward Probabilities:")
print(beta)
```
