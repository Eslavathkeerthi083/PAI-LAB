```python
import heapq

MOVES = [(-1, 0), (1, 0), (0, -1), (0, 1)]


def manhattan_distance(state, goal):
    distance = 0
    for i in range(3):
        for j in range(3):
            value = state[i][j]
            if value != 0:
                for x in range(3):
                    for y in range(3):
                        if goal[x][y] == value:
                            distance += abs(i - x) + abs(j - y)
    return distance


def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j


def get_neighbors(state):
    neighbors = []
    x, y = find_blank(state)

    for dx, dy in MOVES:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [list(row) for row in state]
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            neighbors.append(tuple(map(tuple, new_state)))
    return neighbors


def a_star(initial, goal):
    pq = []
    heapq.heappush(pq, (0, initial))
    visited = set()
    parent = {initial: None}

    g_cost = {initial: 0}

    while pq:
        _, current = heapq.heappop(pq)

        if current == goal:
            return reconstruct_path(parent, goal)

        visited.add(current)

        for neighbor in get_neighbors(current):
            tentative_g = g_cost[current] + 1

            if neighbor not in g_cost or tentative_g < g_cost[neighbor]:
                g_cost[neighbor] = tentative_g
                f = tentative_g + manhattan_distance(neighbor, goal)
                heapq.heappush(pq, (f, neighbor))
                parent[neighbor] = current

    return None


def reconstruct_path(parent, goal):
    path = []
    current = goal
    while current:
        path.append(current)
        current = parent[current]
    return path[::-1]


def print_state(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()


def read_state(name):
    print(f"Enter {name} state (use 0 for blank):")
    state = []
    for _ in range(3):
        state.append(tuple(map(int, input().split())))
    return tuple(state)


# ---------------- DRIVER CODE ----------------
if __name__ == "__main__":
    initial_state = read_state("INITIAL")
    goal_state = read_state("GOAL")

    path = a_star(initial_state, goal_state)

    if path:
        print("\nSolution Found!")
        print("Number of steps:", len(path) - 1)
        print("\nSteps:\n")

        for step, state in enumerate(path):
            print(f"Step {step}:")
            print_state(state)
    else:
        print("\nGoal state is not reachable from initial state.")
```
