```python
import numpy as np

rows=3
cols=3

gamma=0.9
reward=-1

goal=(2,2)

actions={
'UP':(-1,0),
'DOWN':(1,0),
'LEFT':(0,-1),
'RIGHT':(0,1)
}

V=np.zeros((rows,cols))

# VALUE ITERATION


print("VALUE ITERATION\n")

for iteration in range(50):

    newV=np.copy(V)

    for i in range(rows):
        for j in range(cols):

            if (i,j)==goal:
                continue

            values=[]

            for a in actions:

                di,dj=actions[a]

                ni=i+di
                nj=j+dj

                if ni<0 or ni>=rows or nj<0 or nj>=cols:
                    ni,nj=i,j

                val=reward + gamma*V[ni][nj]

                values.append(val)

            newV[i][j]=max(values)

    print("Iteration:",iteration+1)

    for i in range(rows):
        for j in range(cols):
            print("State",(i,j),"Value",round(newV[i][j],2))

    print("-------------------")

    if np.max(abs(newV-V))<0.001:
        break

    V=newV



# POLICY ITERATION


print("\nPOLICY ITERATION\n")

policy=np.full((rows,cols),"DOWN",dtype=object)

p_iteration=1

while True:

    print("Policy Iteration:",p_iteration)

    # Policy Evaluation
    for k in range(30):

        for i in range(rows):
            for j in range(cols):

                if (i,j)==goal:
                    continue

                act=policy[i][j]

                di,dj=actions[act]

                ni=i+di
                nj=j+dj

                if ni<0 or ni>=rows or nj<0 or nj>=cols:
                    ni,nj=i,j

                V[i][j]=reward + gamma*V[ni][nj]

    print("State Values:")

    for i in range(rows):
        for j in range(cols):
            print("State",(i,j),"Value",round(V[i][j],2))

    print("Current Policy:")

    for i in range(rows):
        for j in range(cols):
            print("State",(i,j),"Action",policy[i][j])

    stable=True

    # Policy Improvement
    for i in range(rows):
        for j in range(cols):

            if (i,j)==goal:
                continue

            old=policy[i][j]

            best_action=None
            best_value=-999

            for a in actions:

                di,dj=actions[a]

                ni=i+di
                nj=j+dj

                if ni<0 or ni>=rows or nj<0 or nj>=cols:
                    ni,nj=i,j

                val=reward + gamma*V[ni][nj]

                if val>best_value:
                    best_value=val
                    best_action=a

            policy[i][j]=best_action

            if old!=policy[i][j]:
                stable=False

    print("-------------------")

    if stable:
        break

    p_iteration+=1


print("\nFinal Optimal Policy:\n")

for i in range(rows):
    for j in range(cols):

        if (i,j)==goal:
            print("State",(i,j),"GOAL")

        else:
            print("State",(i,j),"Action",policy[i][j])
```
